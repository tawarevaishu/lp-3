# Assignment 3: Gradient Descent Algorithm

**Aim:** Implement Gradient Descent to find the local minimum of `y = (x + 3)²`, starting from `x = 2`.

This notebook records each iteration, displays the convergence table, and plots the function and gradient descent path.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Define the function and its derivative
Given function: **y = (x + 3)²**. Its derivative is **dy/dx = 2(x + 3)**. Gradient descent updates x in the opposite direction of the derivative.

```python
def function(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)

# Starting point and learning rate
x = 2.0
learning_rate = 0.1
iterations = 50

history = []
```

## 3. Implement Gradient Descent

```python
for i in range(iterations):
    y = function(x)
    grad = derivative(x)
    history.append([i, x, y, grad])
    x = x - learning_rate * grad

history_df = pd.DataFrame(history, columns=["Iteration", "x", "y", "Gradient"])
display(history_df.head(10))
print("Approximate minimum x:", x)
print("Function value at minimum:", function(x))
```

## 4. Plot convergence

```python
history_array = np.array(history)
plt.figure(figsize=(8, 5))
plt.plot(history_array[:, 0], history_array[:, 2], marker="o", markersize=3)
plt.xlabel("Iteration")
plt.ylabel("Function value y")
plt.title("Gradient Descent Convergence")
plt.grid(True)
plt.show()

x_values = np.linspace(-6, 4, 300)
plt.figure(figsize=(8, 5))
plt.plot(x_values, function(x_values), label="y = (x + 3)^2")
plt.scatter(history_array[:, 1], history_array[:, 2], color="red", s=20, label="Gradient descent steps")
plt.scatter([-3], [0], color="green", marker="*", s=150, label="Actual minimum (-3, 0)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent on y = (x + 3)^2")
plt.legend()
plt.grid(True)
plt.show()
```

## 5. Conclusion
Starting from **x = 2**, gradient descent repeatedly updates x using `x = x - learning_rate × derivative`. For this convex quadratic function, it converges towards **x = -3**, where the minimum function value is **y = 0**. The learning rate controls the step size: too large a value may overshoot or diverge, while a very small value converges slowly.
